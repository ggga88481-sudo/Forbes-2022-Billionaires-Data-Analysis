# Introduction

The Forbes World's Billionaires List 2022 is a comprehensive dataset detailing information about the wealthiest individuals globally for the year 2022. It captures key demographic, geographic, and financial characteristics of 2,668 billionaires, including their net worth, age, source of wealth, industry category, country of citizenship, and self-made status. The dataset is derived from Forbes' official annual ranking, providing a structured basis for analyzing patterns of global wealth distribution, gender disparities, and economic trends across various industries and nations.

# Data set content

The current dataset 'Forbes World's Billionaires List 2022' includes the following information after initial cleaning operations:

*   **rank**: The billionaire's rank on the Forbes list.
*   **personName**: The billionaire's name.
*   **age**: The billionaire's age.
*   **finalWorth**: The billionaire's net worth in dollars.
*   **category**: The category to which the source of wealth belongs (e.g., Technology, Finance).
*   **source**: The specific source of the billionaire's wealth (e.g., Tesla, Amazon).
*   **country**: The country where the billionaire resides.
*   **city**: The city where the billionaire resides.
*   **selfMade**: A boolean value (True/False) indicating whether the billionaire is self-made or not.
*   **gender**: The billionaire's gender (Male/Female).

These columns provide a comprehensive overview of the world's wealthiest individuals for 2022 and allow for the analysis of wealth distribution patterns, demographics, and sources of wealth.

```python
for col in df.columns:
  pct_missing = np.mean(df[col].isnull())
  print(f'{col} - {pct_missing :.1%}')
```

```python
for col in df.columns:
  pct_missing = np.mean(df[col].isnull())
  if pct_missing > 0.3 :
    df.drop(columns = col , inplace = True)
    print(f"Droping {col} column as it contains - {pct_missing * 100} of nulls")
```

```python
df.shape
```

```python
for  col in df.columns:
  pct_missing = np.mean(df[col].isnull())
  print(f'{col} - {pct_missing :.1%}')
```

```python
for column in df.columns:
  if df[column].dtype == 'object':
    df[column] = df[column].fillna(df[column].mode()[0])
  else:
    df[column] = df[column].fillna(df[column].mean())
```

```python
top_10_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_countries.index, y=top_10_countries.values, hue=top_10_countries.index, palette='viridis', legend=False)
plt.title('Top 10 Countries by Number of Billionaires')
plt.xlabel('Country')
plt.ylabel('Number of Billionaires')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
